## Data augmentation pipeline

Augmentationt pipelines for Model #1. Each function returns a fastai Transform list that can be passed directly to a DataBlock or DataLoaders.

Relevant augmentations:
- brightness
- rotation
- blur
- scaling
- flipping

Design note: top-down food photography has specific characteristics:
- Rotation up to ±30° (user hand angle variation)
- Moderate blur simulates slight camera shake at distance
- Brightness / contrast variation simulates kitchen lighting changes

# Loading data

Dataset loader for:
- Object detection (COCO JSON bounding-box annotations)

Label Studio COCO export format assumptions:
- Detection export  : standard COCO format with "images", "annotations", "categories" keys. Each annotation has bbox in [x_min, y_min, width, height] format.

Split strategy:
- Splits are performed at the image level (not annotation level) to avoid data leakage between train / val / test.
- Ratios: 80 / 10 / 10
- Random seed is fixed for reproducibility.

## COCO JSON helpers

```python
def load_coco_json(json_path: Path) -> dict:
    """
    Load and validate a COCO JSON file exported from Label Studio.
    """
    if not json_path.exists():
        raise FileNotFoundError(
            f"COCO annotation file not found: {json_path}\n"
            f"Export your Label Studio project as COCO JSON and place it at "
            f"the path defined in {COCO_MODEL1_JSON}."
        )
    with open(json_path, "r") as f:
        data = json.load(f)

    required_keys = {"images", "annotations", "categories"}

    if not required_keys.issubset(data.keys()):
        raise ValueError(
            f"COCO JSON at {json_path} is missing keys: "
            f"{required_keys - data.keys()}"
        )
    return data



def split_image_ids(
        image_ids: list,
        train_ratio: float = SPLIT_TRAIN,
        val_ratio: float = SPLIT_VAL,
        seed: int = RANDOM_SEED
    ) -> tuple:
    """
    Randomly split a list of image IDs into train / val / test sets.

    Returns:
        tuple[set, set, set]
            (train_ids, val_ids, test_ids)
    """
    rng = np.random.default_rng(seed)
    ids = np.array(image_ids)
    rng.shuffle(ids)

    n = len(ids)
    n_train = int(n * train_ratio)
    n_val = int(n * val_ratio)

    train_ids = set(ids[:n_train].tolist())
    val_ids = set(ids[n_train:n_train + n_val].tolist())
    test_ids = set(ids[n_train + n_val:].tolist())

    print(f"Split -> train: {len(train_ids)} | val: {len(val_ids)} | test: {len(test_ids)} images")
    return train_ids, val_ids, test_ids



def build_annotation_df(coco: dict, image_dir: Path) -> tuple:
    """
    Parse a COCO dict into a flat DataFrame with one row per annotation.

    bbox is converted from COCO format [x, y, w, h] to Pascal VOC
    [x_min, y_min, x_max, y_max] as required by torchvision.

    Returns (full_df, lable_to_idx) where label_to_idx maps class name
    to the torchvision label index (1-based, 0 = background).
    """
    # Category mapping - torchvision needs 1-based integer labels
    # Sort by category if for deterministic ordering
    cats_sorted = sorted(coco["categories"], key = lambda c: c["id"])
    label_to_idx = {cat["name"]: i+1 for i, cat in enumerate(cats_sorted)}
    print(f"Label -> index mapping: {label_to_idx}")

    # Warn if COCO labels differ from MODEL1_CLASSES
    coco_labels = set(label_to_idx.keys())
    cfg_labels = set(MODEL1_CLASSES)
    if coco_labels != cfg_labels:
        print(f"WARNING: COCO labels {coco_labels} differ from "
            f"MODEL1_CLASSES {cfg_labels}. Update MODEL1_CLASSES if needed.")

    # Image id -> cleaned filename
    id_to_file = {
        img["id"]: Path(img["file_name"].replace("\\", "/")).name.split("-", 1)[-1]
        for img in coco["images"]
    }

    all_image_ids = list(id_to_file.keys())
    print(f"Total images: {len(all_image_ids)}")

    train_ids, val_ids, test_ids = split_image_ids(all_image_ids)

    rows = []
    for ann in coco["annotations"]:
        img_id = ann["image_id"]
        filename = id_to_file[img_id]
        x, y, w, h = ann["bbox"]
        label_name = {cat["id"]: cat["name"] for cat in coco["categories"]}[
            ann["category_id"]
        ]
        split = (
            "train" if img_id in train_ids else
            "valid" if img_id in val_ids else
            "test"
        )
        rows.append({
            "image_path": str(image_dir / filename),
            "x_min": float(x),
            "y_min": float(y),
            "x_max": float(x + w),
            "y_max": float(y + h),
            "label": label_name,
            "label_idx": label_to_idx[label_name],
            "split": split
        })

    return pd.DataFrame(rows), label_to_idx
```

## Dataset

```python
class GNDetectionDataset(data.Dataset):
    """
    PyTorch Dataset for GN container detection.

    Each item returns:
        image: FloatTensor [3, H, W], pixel values in [0, 1]
        target: dict with keys:
            boxes : FloatTensor [N, 4] - Pascal VOC format
            labels : Int64Tensor [N] - 1-based class indices
            image_id : Int64Tensor [1]
    """
    def __init__(self, df: pd.DataFrame, transforms = None):
        """
        Parameters
            df : pd.DataFrame
                Annotation DataFrame with columns:
                image_path, x_min, y_min, x_max, y_max, label_idx
            transforms : callable, optional
                torchvision transform applied to (image, target) pairs.
        """
        # Group by image so each __getitem__ returns all boxes for one image
        self.records = []
        for img_path, grp in df.groupby("image_path"):
            self.records.append({
                "image_path": img_path,
                "boxes": grp[["x_min", "y_min", "x_max", "y_max"]].values,
                "labels": grp["label_idx"].values
            })
        self.transforms = transforms


    def __len__(self):
        return len(self.records)


    def __getitem__(self, idx):
        rec = self.records[idx]
        image = Image.open(rec["image_path"]).convert("RGB")
        image = TF.to_tensor(image) # FloatTensor [3, H, W] on [0, 1]

        boxes = torch.as_tensor(rec["boxes"], dtype = torch.float32)
        labels = torch.as_tensor(rec["labels"], dtype = torch.int64)

        target = {
            "boxes": boxes,
            "labels": labels,
            "image_ids": torch.tensor([idx])
        }
        if self.transforms:
            image, target = self.transforms(image, target)

        return image, target



def collate_fn(batch):
    """
    Custom collate: Faster R-CNN expects a list of (image, target) pairs,
    not a stacked tensor, because images can have different numbers of boxes.
    """
    return tuple(zip(*batch))
```

## Transforms

```python
class Compose:
    """
    Chain multiple (image, target) transforms.
    """
    def __init__(self, transforms):
        self.transforms = transforms


    def __call__(self, image, target):
        for t in self.transforms:
            image, target = t(image, target)

        return image, target



class RandomHorizontalFlip:
    """
    Horizontally flip image and adjust bounding boxes.
    """
    def __init__(self, prob = 0.5):
        self.prob = prob


    def __call__(self, image, target):
        if torch.rand(1) < self.prob:
            _, h, w = image.shape # C, H, W
            image = TF.hflip(image)
            boxes = target["boxes"].clone()
            boxes[:, [0, 2]] = w - boxes[:, [2,0]]
            target["boxes"] = boxes

        return image, target



class ResizeLongestEdge:
    """
    Resize so that the longest edge equals 'size', preserving aspect ratio.
    """
    def __init__(self, size = IMAGE_SIZE):
        self.size = size


    def __call__(self, image, target):
        _, h, w = image.shape
        scale = self.size / max(h, w)
        new_h = int(h * scale)
        new_w = int(w * scale)
        image = TF.resize(image, [new_h, new_w])

        if target is not None and "boxes" in target:
            target["boxes"] = target["boxes"] * scale

        return image, target


class ColorJitter:
    """
    Apply brightness and contrast jitter.
    """
    def __init__(self, brightness = 0.3, contrast = 0.3):
        self.jitter = T.ColorJitter(brightness = brightness, contrast = contrast)


    def __call__(self, image, target):
        image = self.jitter(image)
        return image, target


class RandomRotation:
    def __init__(self, degrees=30, prob=0.5):
        self.degrees = degrees
        self.prob = prob

    def __call__(self, image, target):
        if torch.rand(1) < self.prob:
            angle = float(torch.empty(1).uniform_(-self.degrees, self.degrees))

            _, h, w = image.shape
            image = TF.rotate(image, angle)

            # --- Rotate bounding boxes ---
            boxes = target["boxes"]
            cx, cy = w / 2, h / 2

            new_boxes = []
            for box in boxes:
                x_min, y_min, x_max, y_max = box

                corners = torch.tensor([
                    [x_min, y_min],
                    [x_max, y_min],
                    [x_max, y_max],
                    [x_min, y_max]
                ])

                # Translate to center
                corners[:, 0] -= cx
                corners[:, 1] -= cy

                # Rotation matrix
                theta = -angle * np.pi / 180
                rot = torch.tensor([
                    [np.cos(theta), -np.sin(theta)],
                    [np.sin(theta),  np.cos(theta)]
                ])

                corners = corners @ rot.T

                # Translate back
                corners[:, 0] += cx
                corners[:, 1] += cy

                x_coords = corners[:, 0]
                y_coords = corners[:, 1]

                new_boxes.append([
                    x_coords.min(),
                    y_coords.min(),
                    x_coords.max(),
                    y_coords.max()
                ])

            target["boxes"] = torch.stack([torch.tensor(b) for b in new_boxes])

        return image, target


class RandomScale:
    def __init__(self, scale_range=(0.8, 1.2), prob=0.5):
        self.scale_range = scale_range
        self.prob = prob

    def __call__(self, image, target):
        if torch.rand(1) < self.prob:
            scale = float(torch.empty(1).uniform_(*self.scale_range))

            _, h, w = image.shape
            new_h, new_w = int(h * scale), int(w * scale)

            image = TF.resize(image, [new_h, new_w])

            target["boxes"] = target["boxes"] * scale

        return image, target


class RandomBlur:
    def __init__(self, kernel_size=5, prob=0.3):
        self.kernel_size = kernel_size
        self.prob = prob

    def __call__(self, image, target):
        if torch.rand(1) < self.prob:
            image = TF.gaussian_blur(image, kernel_size=self.kernel_size)
        return image, target


class RandomVerticalFlip:
    def __init__(self, prob=0.2):
        self.prob = prob

    def __call__(self, image, target):
        if torch.rand(1) < self.prob:
            _, h, w = image.shape
            image = TF.vflip(image)

            boxes = target["boxes"].clone()
            boxes[:, [1, 3]] = h - boxes[:, [3, 1]]
            target["boxes"] = boxes

        return image, target


def get_train_transforms():
    return Compose([
        ResizeLongestEdge(IMAGE_SIZE),
        
        RandomHorizontalFlip(prob = 0.5),
        #RandomVerticalFlip(prob=0.1),

        #RandomRotation(degrees=20, prob=0.4),
        #RandomScale(scale_range=(0.85, 1.15), prob=0.1),
        
        ColorJitter(brightness = 0.2, contrast = 0.2)
        #RandomBlur(kernel_size=5, prob=0.2)
    ])


def get_val_transforms():
    return Compose([
        ResizeLongestEdge(IMAGE_SIZE)
    ])
```

# Model

```python
def build_model(num_classes: int) -> torch.nn.Module:
    """
    Load Faster R-CNN ResNet-50 FPN v2 with ImageNet-pretrained weights,
    then replace the classification head to match our number of classes.

    num_classes includes background(e.g. 3 = background + small + large).
    """
    weights = FasterRCNN_ResNet50_FPN_V2_Weights.DEFAULT
    model = fasterrcnn_resnet50_fpn_v2(weights = weights)

    # Replace the box predictor head - this is the standard torchvision pattern
    in_features = model.roi_heads.box_predictor.cls_score.in_features
    model.roi_heads.box_predictor = FastRCNNPredictor(in_features, num_classes)

    return model
```

```python
print(f"[Model] Building Faster R-CNN ResNet-50 FPN v2"
    f"(num_classes = {NUM_CLASSES} including background).")

model = build_model(NUM_CLASSES)
model.to(DEVICE)
print("Model ready.")
```

## Freeze / unfreeze helpers

```python
def freeze_backbone(model: torch.nn.Module):
    """
    Freeze all backbone and FPN parameters.
    Only the RPN and ROI heads will be trained in Phase 1.
    """
    for name, param in model.named_parameters():
        if "backbone" in name or "fpn" in name:
            param.requires_grad = False
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)

    print(f"Backbone frozen — trainable parameters: {trainable:,}")


def unfreeze_all(model: torch.nn.Module):
    """
    Unfreeze all parameters for Phase 2 fine-tuning.
    """
    for param in model.parameters():
        param.requires_grad = True
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)

    print(f"All layers unfrozen — trainable parameters: {trainable:,}")


def get_optimizer_phase1(model, lr = MODEL1_LR_FROZEN, wd = MODEL1_WEIGHT_DECAY):
    """
    Optimise only trainable (non-frozen) parameters.
    """
    params = [p for p in model.parameters() if p.requires_grad]

    return torch.optim.SGD(params, lr = lr, momentum = 0.9, weight_decay = wd)


def get_optimizer_phase2(model, lr = MODEL1_LR_UNFROZEN, wd = MODEL1_WEIGHT_DECAY):
    """
    Discriminative learning rates for Phase 2:
    backbone gets lr/10, heads get full lr.
    """
    backbone_params = [
        p for n, p in model.named_parameters()
        if ("backbone" in n or "fpn" in n) and p.requires_grad
    ]
    head_params = [
        p for n, p in model.named_parameters()
        if ("backbone" not in n and "fpn" not in n) and p.requires_grad
    ]

    return torch.optim.SGD(
        [
            {"params": backbone_params, "lr": lr / 10},
            {"params": head_params,     "lr": lr},
        ],
        momentum = 0.9,
        weight_decay = wd)
```

# Training loop

```python
def train_one_epoch(model, loader, optimizer, device, scaler = None, epoch = 0):
    """
    One training epoch.

    Faster R-CNN returns a dict of losses when called in training mode:
        loss_classifier   — RoI classification loss
        loss_box_reg      — RoI bounding box regression loss
        loss_objectness   — RPN objectness loss
        loss_rpn_box_reg  — RPN box regression loss
    Total loss is their sum.
    """
    model.train()
    total_loss = 0.0

    for i, (images, targets) in enumerate(loader):
        images  = [img.to(device) for img in images]
        targets = [{k: v.to(device) for k, v in t.items()} for t in targets]

        optimizer.zero_grad()

        if scaler is not None:
            with torch.cuda.amp.autocast():
                loss_dict = model(images, targets)
                loss = sum(loss_dict.values())
            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()
        else:
            loss_dict = model(images, targets)
            loss = sum(loss_dict.values())
            loss.backward()
            optimizer.step()

        total_loss += loss.item()

        if (i + 1) % 5 == 0 or (i + 1) == len(loader):
            print(f"  Epoch {epoch+1} [{i+1}/{len(loader)}]  "
                  f"loss: {loss.item():.4f}  "
                  f"(cls: {loss_dict['loss_classifier'].item():.3f}  "
                  f"box: {loss_dict['loss_box_reg'].item():.3f}  "
                  f"rpn: {loss_dict['loss_objectness'].item():.3f})")

    return total_loss / len(loader)


@torch.no_grad()
def evaluate_one_epoch(model, loader, device):
    """
    Validation epoch.

    Note: Faster R-CNN only returns losses when targets are supplied AND the
    model is in training mode. To get validation loss we temporarily switch
    to train mode — this is the standard torchvision pattern for monitoring
    validation loss without running full COCO mAP evaluation (which requires
    the pycocotools package).
    """
    model.train()   # needed to get loss dict — see note above
    total_loss = 0.0
    with torch.no_grad():
        for images, targets in loader:
            images  = [img.to(device) for img in images]
            targets = [{k: v.to(device) for k, v in t.items()} for t in targets]
            loss_dict  = model(images, targets)
            total_loss += sum(loss_dict.values()).item()
    model.eval()

    return total_loss / len(loader)


def run_training(
    model,
    train_loader,
    val_loader,
    optimizer,
    scheduler,
    n_epochs,
    device,
    scaler = None,
    phase_label = "Phase"
):
    """
    Training loop with early stopping and best-model checkpointing.

    Returns the best validation loss achieved.
    """
    best_val_loss  = float("inf")
    best_weights   = copy.deepcopy(model.state_dict())
    patience_count = 0
    log_rows       = []

    for epoch in range(n_epochs):
        t0 = time.time()

        train_loss = train_one_epoch(
            model, train_loader, optimizer, device, scaler, epoch
        )
        val_loss = evaluate_one_epoch(model, val_loader, device)

        if scheduler is not None:
            scheduler.step()

        elapsed = time.time() - t0
        print(f"[{phase_label}] Epoch {epoch+1}/{n_epochs}  "
              f"train_loss: {train_loss:.4f}  val_loss: {val_loss:.4f}  "
              f"({elapsed:.0f}s)")

        log_rows.append({
            "phase": phase_label, "epoch": epoch + 1,
            "train_loss": train_loss, "val_loss": val_loss,
        })

        # Best model checkpoint
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_weights  = copy.deepcopy(model.state_dict())
            patience_count = 0
            print(f"  ✓ New best val_loss: {best_val_loss:.4f} — checkpoint saved")
        else:
            patience_count += 1
            print(f"  No improvement ({patience_count}/{EARLY_STOPPING_PATIENCE})")
            if patience_count >= EARLY_STOPPING_PATIENCE:
                print(f"  Early stopping triggered after epoch {epoch+1}.")
                break

    # Restore best weights
    model.load_state_dict(best_weights)

    return best_val_loss, pd.DataFrame(log_rows)
```

```python
print(f"\n{'='*60}")
print(f"  Phase 1 — frozen backbone ({MODEL1_EPOCHS_FROZEN} epochs)")
print(f"{'='*60}")

freeze_backbone(model)
optimizer_p1 = get_optimizer_phase1(model)
scheduler_p1 = torch.optim.lr_scheduler.OneCycleLR(
    optimizer_p1,
    max_lr = MODEL1_LR_FROZEN,
    steps_per_epoch = len(train_loader),
    epochs = MODEL1_EPOCHS_FROZEN,
    pct_start = 0.3
)

best_loss_p1, log_p1 = run_training(
    model, train_loader, val_loader,
    optimizer_p1, scheduler_p1,
    MODEL1_EPOCHS_FROZEN, DEVICE, scaler,
    phase_label = "Phase1",
)

print(f"\nPhase 1 complete — best val_loss: {best_loss_p1:.4f}")
```

## Phase 2 - full fine-tune

```python
print(f"\n{'='*60}")
print(f"  Phase 2 — full fine-tune ({MODEL1_EPOCHS_UNFROZEN} epochs)")
print(f"{'='*60}")

unfreeze_all(model)
optimizer_p2 = get_optimizer_phase2(model)
scheduler_p2 = torch.optim.lr_scheduler.OneCycleLR(
    optimizer_p2,
    max_lr = [MODEL1_LR_UNFROZEN / 10, MODEL1_LR_UNFROZEN],
    steps_per_epoch = len(train_loader),
    epochs = MODEL1_EPOCHS_UNFROZEN,
    pct_start = 0.3,
)

best_loss_p2, log_p2 = run_training(
    model, train_loader, val_loader,
    optimizer_p2, scheduler_p2,
    MODEL1_EPOCHS_UNFROZEN, DEVICE, scaler,
    phase_label = "Phase2",
)

print(f"\nPhase 2 complete — best val_loss: {best_loss_p2:.4f}")
```

# Visualisation

```python
# =============================================================================
# Visualisation — draw predicted and ground-truth boxes on validation images
# =============================================================================
# Add this as a new cell at the bottom of your Colab notebook.
# Run after the inference check cell.
#
# Controls:
#   SCORE_THRESHOLD  — only show predictions above this confidence score.
#                      Start at 0.05 (low, to see everything), raise it
#                      once the model improves (e.g. 0.5 in production).
#   NUM_IMAGES       — how many validation images to display.
# =============================================================================

import matplotlib.pyplot as plt
import matplotlib.patches as patches
import matplotlib.patheffects as pe


# --- Settings ----------------------------------------------------------------

SCORE_THRESHOLD = 0.05   # lower = show more boxes; raise as model improves
NUM_IMAGES      = 4      # number of val images to display


# --- Colour map --------------------------------------------------------------
# Index 0 = background (never shown), 1 and 2 = your two container classes.
# Ground-truth boxes are drawn in a lighter tint of the same colour.

CLASS_COLOURS = {
    0: "#888888",   # background — not drawn
    1: "#E8593C",   # large_container  — coral/orange
    2: "#1D9E75",   # small_container  — teal/green
}


# --- Helper ------------------------------------------------------------------

def draw_boxes(ax, boxes, labels, scores=None, class_colours=CLASS_COLOURS,
               idx_to_label=None, is_gt=False):
    """
    Draw bounding boxes on a matplotlib Axes.

    Parameters
    ----------
    ax           : matplotlib Axes
    boxes        : list of [x_min, y_min, x_max, y_max]
    labels       : list of int (1-based class indices)
    scores       : list of float, or None (ground truth has no scores)
    class_colours: dict {label_idx: hex_colour}
    idx_to_label : dict {label_idx: label_name}
    is_gt        : bool — True draws dashed border (ground truth style)
    """
    for i, (box, label) in enumerate(zip(boxes, labels)):
        x_min, y_min, x_max, y_max = box
        w = x_max - x_min
        h = y_max - y_min

        colour    = class_colours.get(int(label), "#888888")
        linestyle = "--" if is_gt else "-"
        linewidth = 1.5 if is_gt else 2.0
        alpha     = 0.6 if is_gt else 0.9

        rect = patches.FancyBboxPatch(
            (x_min, y_min), w, h,
            boxstyle="square,pad=0",
            linewidth=linewidth,
            edgecolor=colour,
            facecolor="none",
            linestyle=linestyle,
            alpha=alpha,
        )
        ax.add_patch(rect)

        # Label text
        label_name = idx_to_label.get(int(label), str(label)) if idx_to_label else str(label)
        if is_gt:
            text = f"GT: {label_name}"
        elif scores is not None:
            text = f"{label_name} {scores[i]:.2f}"
        else:
            text = label_name

        txt = ax.text(
            x_min + 3, y_min + 3,
            text,
            fontsize=7,
            color="white",
            verticalalignment="top",
        )
        # Dark outline so text is readable on any background
        txt.set_path_effects([
            pe.Stroke(linewidth=2, foreground=colour),
            pe.Normal(),
        ])


# --- Main visualisation loop -------------------------------------------------

model.eval()

# Pick up to NUM_IMAGES images from the validation set
n_show  = min(NUM_IMAGES, len(val_dataset))
indices = list(range(n_show))

fig, axes = plt.subplots(
    n_show, 1,
    figsize=(10, 6 * n_show),
    squeeze=False,
)

for plot_idx, sample_idx in enumerate(indices):
    image_tensor, gt_target = val_dataset[sample_idx]
    record = val_dataset.records[sample_idx]

    # Run inference
    with torch.no_grad():
        predictions = model([image_tensor.to(DEVICE)])[0]

    # Convert image tensor [C, H, W] → numpy [H, W, C] for matplotlib
    img_np = image_tensor.permute(1, 2, 0).cpu().numpy()
    img_np = img_np.clip(0, 1)

    ax = axes[plot_idx][0]
    ax.imshow(img_np)
    ax.set_axis_off()

    filename = Path(record["image_path"]).name
    ax.set_title(
        f"{filename}",
        fontsize=9, pad=4, loc="left",
    )

    # Filter predictions by score threshold
    pred_boxes  = predictions["boxes"].cpu().tolist()
    pred_labels = predictions["labels"].cpu().tolist()
    pred_scores = predictions["scores"].cpu().tolist()

    keep = [i for i, s in enumerate(pred_scores) if s >= SCORE_THRESHOLD]
    pred_boxes_f  = [pred_boxes[i]  for i in keep]
    pred_labels_f = [pred_labels[i] for i in keep]
    pred_scores_f = [pred_scores[i] for i in keep]

    # Ground truth boxes
    gt_boxes  = gt_target["boxes"].tolist()
    gt_labels = gt_target["labels"].tolist()

    # Draw ground truth first (dashed), then predictions (solid) on top
    draw_boxes(ax, gt_boxes,      gt_labels,      scores=None,
               idx_to_label=idx_to_label, is_gt=True)
    draw_boxes(ax, pred_boxes_f,  pred_labels_f,  scores=pred_scores_f,
               idx_to_label=idx_to_label, is_gt=False)

    # Legend summary
    n_pred = len(pred_boxes_f)
    n_gt   = len(gt_boxes)
    ax.set_xlabel(
        f"Ground truth: {n_gt} box(es)   |   "
        f"Predicted (score ≥ {SCORE_THRESHOLD}): {n_pred} box(es)",
        fontsize=8, labelpad=4,
    )

plt.suptitle(
    f"Model #1 — Faster R-CNN predictions\n"
    f"Coral = large_container  |  Teal = small_container  |  "
    f"Dashed = ground truth  |  Solid = prediction",
    fontsize=9, y=1.002,
)
plt.tight_layout()
plt.show()

print(f"\nShowing {n_show} validation image(s).")
print(f"Score threshold: {SCORE_THRESHOLD}  "
      f"(raise this as training improves — use 0.4–0.6 in production)")
```
